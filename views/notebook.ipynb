```python
import requests
import json

def fetch_products(url):
    response = requests.get(url)
    response.raise_for_status()
    return response.json()

def parse_products(data):
    headers = list(data["headers"].keys())
    products = []

    for product_array in data["products"]:
        product = {}
        for i, key in enumerate(headers):
            if i < len(product_array):
                product[key] = product_array[i]
            else:
                product[key] = None
        products.append(product)

    return products

def print_product(product):
    print("-" * 60)
    for key, value in product.items():
        print(f"  {key}: {value}")
    print()


URL = "https://www.diodes.com/productcollection/json/9695/collection?eid=913&bid=10020"

data = fetch_products(URL)
products = parse_products(data)

print(f"Found {len(products)} products\n")

for product in products:
    print_product(product)

automotive = [
    p for p in products
    if "Automotive" in (p.get("Compliance (Only Automotive Supports PPAP)") or [])
]
print(f"\nAutomotive parts: {len(automotive)}")

unipolar = [
    p for p in products
    if "Unipolar" in (p.get("Type") or [])
]
print(f"Unipolar parts: {len(unipolar)}")

print("\nAll part numbers:")
for p in products:
    print(f"  {p['PartNumber']} - {p['Description']}")
```
